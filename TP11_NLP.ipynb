```python

text = "The CATS are RUNNING! They love playing, and it's amazing @2026."

print("Texte original :", text)
```

```python

text_minuscule = text.lower()

print("Texte en minuscules :", text_minuscule)
```

```python

text_nettoye = re.sub(r'[^a-z\s]', ' ', text_minuscule)
text_nettoye = re.sub(r'\s+', ' ', text_nettoye).strip()

print("Texte après nettoyage :", text_nettoye)
```

```python

tokens = word_tokenize(text_nettoye)

print("Liste des tokens :", tokens)
```

```python

stop_words = set(stopwords.words('english'))

print("Liste des stopwords :")
print(sorted(stop_words))

tokens_filtres = [mot for mot in tokens if mot not in stop_words]

print("\nMots après suppression des stopwords :")
print(tokens_filtres)
```

```python

mot_a_ajouter = "amazing"
stop_words.add(mot_a_ajouter)

tokens_filtres = [mot for mot in tokens if mot not in stop_words]

print("Mots restants :", tokens_filtres)
```

```python

nombre_initial = len(tokens)
nombre_restant = len(tokens_filtres)

if nombre_initial > 0:
    proportion = ((nombre_initial - nombre_restant) / nombre_initial) * 100
else:
    proportion = 0

print("Nombre initial de mots :", nombre_initial)
print("Nombre de mots restants :", nombre_restant)
print("Proportion supprimée :", round(proportion, 2), "%")
```

```python

lemmatizer = WordNetLemmatizer()

tokens_lemmatises = [
    lemmatizer.lemmatize(mot, pos='v')
    for mot in tokens_filtres
]

print("Avant lemmatisation :", tokens_filtres)
print("Après lemmatisation :", tokens_lemmatises)
```

```python

doc = nlp("The cats are running and playing.")

for token in doc:
    print(token.text, "->", token.lemma_)
```

```python

phrases = [
    "I love football",
    "I love cinema",
    "Football is a popular sport"
]

vectorizer = CountVectorizer()
matrice_bow = vectorizer.fit_transform(phrases)

print("Vocabulaire :")
print(vectorizer.get_feature_names_out())

print("\nMatrice Bag of Words :")
print(matrice_bow.toarray())
```

```python

tfidf_vectorizer = TfidfVectorizer()
matrice_tfidf = tfidf_vectorizer.fit_transform(phrases)

print("Vocabulaire TF-IDF :")
print(tfidf_vectorizer.get_feature_names_out())

print("\nMatrice TF-IDF :")
print(matrice_tfidf.toarray())
```

```python

for i, phrase in enumerate(phrases, start=1):
    print(f"Phrase {i} : {phrase}")
    print("Nombre de mots :", len(phrase.split()))
    print()
```
